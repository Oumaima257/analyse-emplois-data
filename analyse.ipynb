```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("ds_salaries.csv")
df.head()
```

```python
print(df.shape)
df.info()
print("Doublons :", df.duplicated().sum())
```

```python
# 1. Supprimer la colonne inutile
df = df.drop(columns=["Unnamed: 0"])

# 2. Remplacer les codes par des libellés lisibles
df["experience_level"] = df["experience_level"].map({
    "EN": "Débutant",
    "MI": "Intermédiaire",
    "SE": "Senior",
    "EX": "Direction"
})

df["company_size"] = df["company_size"].map({
    "S": "Petite",
    "M": "Moyenne",
    "L": "Grande"
})

df["employment_type"] = df["employment_type"].map({
    "FT": "Temps plein",
    "PT": "Temps partiel",
    "CT": "Contrat",
    "FL": "Freelance"
})

# 3. Vérifier le résultat
df.head()
```

```python
# Salaire moyen par niveau d'expérience
df.groupby("experience_level")["salary_in_usd"].mean().sort_values()
```

```python
ordre = ["Débutant", "Intermédiaire", "Senior", "Direction"]
sns.barplot(data=df, x="experience_level", y="salary_in_usd", order=ordre)
plt.title("Salaire moyen selon le niveau d'expérience")
plt.xlabel("Niveau d'expérience")
plt.ylabel("Salaire (USD)")
plt.show()
df.groupby("experience_level")["salary_in_usd"].agg(["mean", "count"]).round(0)
```

# **Le salaire augmente avec l'experience, de X USD pour les débutants à Y USD pour la direction. Le groupe « Direction » compte peu de profils (N), donc sa moyenne est moins fiable.**

```python
top_jobs = (df.groupby("job_title")["salary_in_usd"]
              .agg(["mean", "count"])
              .query("count >= 5")
              .sort_values("mean", ascending=False)
              .head(10))
top_jobs
```

```python
top_jobs["mean"].sort_values().plot(kind="barh", figsize=(8, 5))
plt.title("Top 10 des métiers data les mieux payés (salaire moyen en USD)")
plt.xlabel("Salaire moyen (USD)")
plt.tight_layout()
plt.show()
```

```python
# Salaire moyen par pays de l'entreprise (au moins 10 offres)
par_pays = (df.groupby("company_location")["salary_in_usd"]
              .agg(["mean", "count"])
              .query("count >= 10")
              .sort_values("mean", ascending=False))
par_pays
```

```python
par_pays["mean"].sort_values().plot(kind="barh", figsize=(8, 5))
plt.title("Salaire moyen par pays (au moins 10 offres)")
plt.xlabel("Salaire moyen (USD)")
plt.tight_layout()
plt.show()
```
